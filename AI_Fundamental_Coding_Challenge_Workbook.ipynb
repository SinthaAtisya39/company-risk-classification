## Challenge 3 - Prediction

**Tujuan:** Menggunakan model yang telah dilatih untuk memprediksi data baru yang belum pernah dilihat.

### Instruksi
- Muat kembali model yang sudah disimpan di Challenge 2
- Buat 2-3 contoh data baru dengan fitur yang sama dengan X
- Lakukan prediksi dan tampilkan hasilnya; jika model mendukung `predict_proba`, tampilkan probability/confidence scores

```python
# Soal 3.1: Muat model (jika perlu, jika di-run terpisah dari Challenge 2)
import joblib
model = joblib.load('../src/model.pkl')
print('✓ Model loaded')
```

```python
# Soal 3.2: Buat data baru untuk diprediksi
# Data harus memiliki struktur X yang sama (same columns, same order)
# Contoh:

data_baru = pd.DataFrame({
    'kolom1': [nilai1, nilai2, nilai3],
    'kolom2': [nilai1, nilai2, nilai3],
    # ... tambahkan semua kolom yang ada di X ...
})

print('Data baru untuk diprediksi:')
print(data_baru)
```

```python
# Soal 3.3: Lakukan prediksi dan tampilkan hasil

hasil_prediksi = model.predict(data_baru)
print(f'Hasil Prediksi: {hasil_prediksi}')

# Bonus: Tampilkan probability jika model mendukung predict_proba.
# Untuk multiclass, akan ditampilkan probability untuk semua kelas.
try:
    hasil_proba = model.predict_proba(data_baru)

    print('\nPrediction Probability:')
    if hasattr(model, 'classes_'):
        kelas = model.classes_
        for i, proba in enumerate(hasil_proba):
            print(f'Sample {i+1}:')
            for label, probability in zip(kelas, proba):
                print(f'  Class {label}: {probability:.2%}')
    else:
        print(hasil_proba)

except AttributeError:
    print('\nModel ini tidak mendukung predict_proba (misalnya beberapa model regression).')
```

## Challenge 4 - Integrasi Gemini API

**Tujuan:** Menggunakan Gemini API untuk menjelaskan hasil prediksi dengan bahasa natural yang mudah dipahami.
- Install Google AI library: `pip install google-generativeai python-dotenv`

```python
# Soal 4.1: Setup Gemini API
#
# OPTION A: Gunakan helper function (jika ada di src/)
# from src.gemini_helper import ask_gemini
#
# OPTION B: Gunakan Google AI library langsung
import google.generativeai as genai
load_dotenv()
API_KEY = os.getenv("GEMINI_API_KEY")

if not API_KEY:
    print("⚠️ ERROR: GEMINI_API_KEY tidak ditemukan!")
    print("Pastikan file .env ada di folder root dengan format: GEMINI_API_KEY=your_key")
else:
    genai.configure(api_key=API_KEY)
    model_gemini = genai.GenerativeModel("gemini-pro")
    print("✓ Gemini API configured successfully")
```

```python
# Soal 4.2: Susun prompt dan minta penjelasan dari Gemini
#
# TIPS: Susun prompt yang menjelaskan:
# 1. Konteks data (apa yang diprediksi)
# 2. Hasil prediksi (apa yang keluar dari model)
# 3. Minta Gemini untuk menjelaskan dalam bahasa mudah
#
# Contoh prompt:

prompt = f"""
Model machine learning telah memprediksi beberapa sampel data baru.

Data baru yang diprediksi:
{data_baru.to_string()}

Hasil prediksi dari model:
{hasil_prediksi}

Penjelasan fitur:
- Kolom1: [jelaskan apa]
- Kolom2: [jelaskan apa]

Pertanyaan: Jelaskan dengan bahasa yang mudah dipahami (bukan teknis) mengapa model
memprediksi hasil tersebut berdasarkan nilai-nilai input yang ada.
Jelaskan juga implikasi atau interpretasi dari hasil prediksi ini dalam konteks nyata.
"""

try:
    response = model_gemini.generate_content(prompt)
    penjelasan = response.text
    print("\n=== Penjelasan dari Gemini ===")
    print(penjelasan)
except Exception as e:
    print(f"Error: {e}")
```
